# 1.2 $\:$ Notation and Background
**Notation.** $a_n \asymp b_n$ means that $a_n / b_n$ and $b_n / a_n$ are bounded for sufficiently large $n$

**Statistical concepts.** Let $\F = \{f(x;\theta) : \theta \in \Theta \}$ be a parametric model (satisfying appropriate regularity conditions).

*Likelihood function.* Based on IID observations $X_1, \dots, X_n$:
$ \L_n(\theta) = \prod_{i=1}^n f(X_i; \theta).$

*Log-likelihood function.*
$ \ell_n(\theta) = \log \L_n(\theta). $

*Maximum likelihood estimator.* $\operatorname{MLE} \hat\theta_n$ is the value of $\theta$ that maximizes the likelihood.

*Score function.* $s(x;\theta) = \frac \partial{\partial \theta} \log f(x;\theta)$ (partial derivative of the log density with respect to theta). It has zero expectation: $\E_\theta (s(X;\theta)) = \int s(x;\theta)f(x;\theta)dx=0$.

The score function appears in the variance of the MLE estimator for $\theta$.

*Fisher information.* The MLE estimator for $\theta$ is asymptotically normal:
$$ \sqrt n (\hat \theta_n - \theta) \rightsquigarrow N(0, \tau^2(\theta)),$$
where $\tau^2(\theta) = 1/I(\theta)$ and
$$I(\theta) = \var_\theta s(X;\theta)
= \E_\theta (s^2(X;\theta)) = -\E_\theta
\left( \frac{\partial^2}{\partial\theta^2} \log f(x;\theta)\right).
$$
is the Fisher information.

We also have
$$ \frac{\hat\theta_n - \theta}{\hat \se} \rightsquigarrow N(0,1),$$
where $\hat\se^2 = 1 / (n I(\hat\theta_n))$, evaluating the Fisher information at the MLE. The Fisher information from $n$ observations satisfies $I_n(\theta) = nI(\theta)$, so we may write $\hat\se^2 = 1 / I_n(\hat\theta_n)$.




# 1.3 $\:$ Confidence Sets

Let $\F$ be a family of distributions and $\theta = \theta(F)$ for $F \in \F$. Let $C_n$ be a set which depends on IID samples $X_1,\dots,X_n$.

**Finite sample $1-\alpha$ confidence set.** The terminology is because this is an exact confidence set obtained from a finite number of samples. $\inf_{F \in \F} \P_F(\theta \in C_n) \ge 1-\alpha.$

**Uniform asymptotic $1-\alpha$ confidence set.** $\liminf_{n\to\infty} \inf_{F\in\F} \P_F(\theta\in C_n) \ge 1-\alpha.$

**Pointwise asympotic $1-\alpha$ confidence set.** For all $F\in \F$, we have $\liminf_{n\to \infty} \P_F(\theta \in C_n) \ge 1-\alpha$.

**Confidence ball.** A confidence set of the form $\{ f\in\F : \lVert f - \hat f_n \rVert\} \le s_n.$

**Confidence envelope.** Also known as a confidence band.
$$\inf_{f\in\F} \P(\ell(x) \le f(x) \le u(x) \text{ for all }x)\ge 1-\alpha.$$

**Example.** Let $\F = \{ f(x;\theta) : \theta \in \Theta \}$ be a parametric model. Let $\hat\theta_n$ be the MLE estimator for $\theta$. We have $\hat \theta_n \approx N(\theta, \hat \se^2)$, where $\hat \se^2 = 1/I_n(\hat\theta_n)^2$. Then $\hat\theta_n \pm z_{\alpha/2}\hat\se$ is a pointwise asympotic confidence interval for $\theta$.

Suppose that $\tau = g(\theta)$. Using the delta method, $\hat\tau_n \approx N(g(\hat\theta_n, |g'(\hat\theta_n))|^2\hat\se^2(\hat\theta_n)).$ A pointwise asympototic confidence interval for $\tau$ is $\hat\tau_n \pm z_{\alpha/2} |g'(\hat\theta_n)|\hat\se(\hat\theta_n).$

# 1.4 $\:$ Useful Inequalities

**Bernstein's inequality.** This is an inequality concerning the size of the sum of bounded random variables with zero mean. Suppose that $X_1,\dots, X_n$ are independent (not necessarily IID), $-M \le X_i \le M$, and $\E X_i = 0$. Then
$$\P\left ( \left | \sum_{i=1}^n X_i \right |  > t \right ) \le 2 \exp
\left ( -\frac 12 \left (\frac{t^2}{v + Mt/3} \right )\right ) $$
for $v \ge \sum_{i=1}^n \var X_i.$

**Bernstein's inequality (moment version).** Suppose that $X_1,\dots, X_n$ are independent (not necessarily IID) with $\E X_i = 0$. Assume
$$ \E |X_i|^m \le \frac{m! M^{m-2}v_i}{2} $$
for all $m\ge 2$. Then
$$\P\left ( \left | \sum_{i=1}^n X_i \right |  > t \right ) \le 2 \exp
\left ( -\frac 12 \left (\frac{t^2}{v + Mt} \right )\right ), $$
where $v = \sum_{i=1}^n v_i.$

**Exercise 1.1.** If $X_1, \dots, X_n \sim \Ber(p)$ are IID, then a pointwise asympotic confidence interval for $p$ is
$$ \hat p_n \pm z_{\alpha/2} \sqrt{\frac{\hat p_n (1-\hat p_n)}{n}} $$
and a uniform confidence interval (obtained using Hoeffding's inequality) is
$$ \hat p_n \pm \sqrt{\frac{1}{2n} \log \left( \frac{2}{\alpha} \right)}.$$

```python
import numpy as np
rng = np.random.default_rng(0)
from scipy.stats import norm
```

```python
# Exercise 1.2
p = 0.20
alpha = 0.05

for n in [10, 100, 1000, 10_000]:
    normal_hits = 0
    hoeffding_hits=0
    for _ in range(10_000):
        X = rng.binomial(n=1,p=p,size=n)
        hat_p = X.mean()

        normal = norm.ppf(1-alpha/2)*np.sqrt(hat_p * (1-hat_p) / n)
        hoeffding = np.sqrt( np.log(2 / alpha) / (2*n) )

        if np.abs(hat_p - p) <= normal:
            normal_hits+=1

        if np.abs(hat_p - p) <= hoeffding:
            hoeffding_hits+=1
    print(f'With {n} samples, the normal approximation confidence interval contains p {normal_hits/10_000:.3f} of the time.')
    print(f'The Hoeffding confidence interval contains p {hoeffding_hits/10_000:.3f} of the time.')
    print()
```

```python
n = 1000
normal = norm.ppf(1-alpha/2)*np.sqrt(hat_p * (1-hat_p) / n)
hoeffding = np.sqrt( np.log(2 / alpha) / (2*n) )
print(2*normal, 2*hoeffding)
```

```python
# At n = 1000 the pointwise interval begins to have accurate coverage.
# The pointwise interval has length 0.049 and the Hoeffding interval has length 0.086.
```

(b) To show that $C_n$ is not a uniform asymptotic confidence interval, we construct $F_n$ such that the corresponding $C_n$ contains $\mu(F_n)$ with probability $\to 0.$

Let $a_n$ be such that $(1-a_n)^n = 1 - \frac 1n$. Then let $F_n$ be such that $ \frac{1}{a_n^2} Y_n \sim F_n$ where $Y_n \sim \Ber(a_n)$. If $X_1,\dots, X_n \sim F_n$ are IID, then $\P(\text{all }X_i\text{ are }0) = 1 - \frac 1n$, so
$$ \P(C_n = \{ 0 \}) = 1 - \frac{1}{n}. $$
On the other hand,
$$ \mu(F_n) = \frac{1}{a_n}.$$

So
$$\P(\mu(F_n) \in C_n) \le \frac{1}{n}.$$

(c) Suppose that $|X_i| \le B$. If $Y_i = X_i - \mu$, then $|Y_i| \le 2B$ and $\var Y_i = \var X_i \le \E |X_i|^2 \le B^2$. By Bernstein's inequality,

$$ \P \left (\left |\sum_{i=1}^n Y_i \right | > t\right) \le 2 \exp\left ( -\frac 12 \left ( \frac{t^2}{n B^2 + 2Bt/3} \right) \right).$$

The left-hand side is
$$ \P( | \bar X_n - \mu | > t/n ), $$
so
$$ \P \left (\left |\bar X_n - \mu \right | > t\right) \le 2 \exp\left ( -\frac 12 \left ( \frac{n^2t^2 }{n B^2 + 2Bnt/3} \right) \right).$$
